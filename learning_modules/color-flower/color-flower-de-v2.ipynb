```python
#teacher-only
from pytamaro.de import Grafik, hsl_farbe, ellipse, fixiere, mitte_links, drehe, zeige_grafik

def blatt(laenge: float, winkel: float) -> Grafik:
    return drehe(winkel, fixiere(mitte_links, ellipse(laenge, laenge / 2, hsl_farbe(winkel, 1, 0.5, 0.5))))

# rotes Blatt (Winkel = Farbton = 0)
zeige_grafik(blatt(200, 0))
```

```python
#teacher-only
from pytamaro.de import leere_grafik, kombiniere, speichere_grafik

def blume(radius: float) -> Grafik:
    result = leere_grafik()
    for winkel in range(0, 360, 30):
        result = kombiniere(result, blatt(radius, winkel))
    return result

zeige_grafik(blume(200))

speichere_grafik("color-flower.svg", blume(200))
```
