```python
#teacher-only
from pytamaro import Graphic, Color, red, green, blue, beside, show_graphic


def rgb_dots(circle_diameter: float) -> Graphic:
    red_dot = circle(circle_diameter, red)
    green_dot = circle(circle_diameter, green)
    blue_dot = circle(circle_diameter, blue)
    return beside(red_dot, beside(green_dot, blue_dot))


show_graphic(rgb_dots(120))
```

```python
#teacher-only
from pytamaro import cyan, magenta, yellow, beside, show_graphic


def cmy_squares(square_side: float) -> Graphic:
    cyan_square = square(square_side, cyan)
    magenta_square = square(square_side, magenta)
    yellow_square = square(square_side, yellow)
    return beside(cyan_square, beside(magenta_square, yellow_square))


show_graphic(cmy_squares(120))
```
