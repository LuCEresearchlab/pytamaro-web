```python
#teacher-only
from pytamaro import (
    Graphic, 
    red,
    triangle,
    compose, pin,
    top_center, top_left, top_right, bottom_center,
    show_graphic, save_graphic,
)

COLOR = red

def sierpinski_triangle(size: int, limit: int) -> Graphic:
    if size <= limit:
        return triangle(size, size, 60, COLOR)
    else:
        smaller = sierpinski_triangle(size // 2, limit)
        return compose(
            pin(top_center, compose(
        pin(top_left, smaller),
        pin(top_right, smaller)
      )),
      pin(bottom_center, smaller)
    )

show_graphic(sierpinski_triangle(256, 16))
save_graphic("sierpinski-triangle.svg", sierpinski_triangle(1024, 32))
```

```python
#teacher-only
from pytamaro import empty_graphic, beside

def evolving_triangles(size: int, limit: int) -> Graphic:
    result = empty_graphic()
    l = size
    while l >= limit:
        result = beside(result, sierpinski_triangle(size, l))
        l /= 2
    return result

show_graphic(evolving_triangles(128, 16))
```
