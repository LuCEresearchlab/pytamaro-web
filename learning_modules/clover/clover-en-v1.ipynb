```python
#teacher-only
from pytamaro import(
    Graphic, Color,
    rgb_color,
    circular_sector, rectangle,
    rotate, pin, compose, above, beside,
    bottom_left, bottom_right, top_left, top_right,
    show_graphic
)


LEAF_COLOR = rgb_color(51, 102, 45)


def semicircle(diameter: float, color: Color) -> Graphic:
    return circular_sector(diameter / 2, 180, color)

def heart(size: float, color: Color) -> Graphic:
    ventricles = rectangle(size, size, color)
    atrium = semicircle(size, color)
    rotated_heart = compose(pin(bottom_right, above(atrium, ventricles)), pin(bottom_left, rotate(-90, atrium)))
    return rotate(45, rotated_heart)

def leaf(size: float) -> Graphic:
    return heart(size, LEAF_COLOR)


show_graphic(leaf(60))
```

```python
#teacher-only
from pytamaro import Graphic, empty_graphic, pin, compose, bottom_center, top_center


def clover(leaf_radius: float, four_leaves: bool) -> Graphic:
    leaf_num = 4 if four_leaves else 3
    rotation_angle = 360 / leaf_num
    result = empty_graphic()
    for i in range(leaf_num):
        result = compose(result, rotate(rotation_angle * i, pin(bottom_center, leaf(leaf_radius))))
    return compose(pin(top_center, stem(leaf_radius)), result)

```
