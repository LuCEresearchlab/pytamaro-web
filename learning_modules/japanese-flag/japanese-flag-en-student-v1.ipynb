```python
from pytamaro import(
    Graphic,
    Color,
    rgb_color,
    rectangle,
    ellipse,
    overlay,
    white,
    show_graphic)


JAPANESE_RED = rgb_color(237, 27, 46)


def japanese_flag(height: float) -> Graphic:
    assert height >= 0
    diameter = height / 5 * 3
    return overlay(
        ellipse(diameter, diameter, JAPANESE_RED),
        rectangle(height / 2 * 3, height, white)
    )


# code to show result
show_graphic(japanese_flag(190))
```
