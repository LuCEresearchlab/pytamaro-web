```python
#teacher-only
from pytamaro import Graphic, Color, triangle, rotate, above, show_graphic


def rhombus(side: float, angle: float, color: Color) -> Graphic:
    assert side > 0
    assert angle < 180
    top_triangle = rotate(-90 - angle / 2, triangle(side, side, angle, color))
    return above(top_triangle, rotate(180, top_triangle))
```

```python
#teacher-only
from pytamaro import Graphic, black, text, beside, above, show_graphic


def double_black_diamond(side: float) -> Graphic:
    diamonds = beside(rhombus(side, 70, black), rhombus(side, 70, black))
    return above(diamonds, text("EXPERTS ONLY", "Helvetica", side / 2, black))


show_graphic(double_black_diamond(120))
```
