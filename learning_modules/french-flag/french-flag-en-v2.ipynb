```python
#teacher-only
from pytamaro import Graphic, Color, rgb_color, white, beside, show_graphic


FRENCH_BLUE = rgb_color(0, 85, 164)
FRENCH_RED = rgb_color(239, 65, 53)


def french_flag(height: float) -> Graphic:
    blue_stripe = stripe(height, FRENCH_BLUE)
    white_stripe = stripe(height, white)
    red_stripe = stripe(height, FRENCH_RED)
    return beside(blue_stripe, beside(white_stripe, red_stripe))


show_graphic(french_flag(300))
```
