```python
#teacher-only
from pytamaro import Color, Graphic
from pytamaro import rectangle, circular_sector, rotate, above, beside, hsl_color, show_graphic

HUE = 229
SATURATION = 1.0

def petal(size: float, color: Color) -> Graphic:
    return above(
        beside(
            rectangle(size, size, color),
            circular_sector(size, 90, color)
        ),
        beside(
            rotate(180, circular_sector(size, 90, color)),
            rectangle(size, size, color)
        )
    )

show_graphic(petal(50, hsl_color(HUE, SATURATION, 0.5)))
```

```python
#teacher-only
from pytamaro import pin, top_left, compose, empty_graphic

def ring(size: float, line_size: float, color: Color) -> Graphic:
    result = empty_graphic()
    for angle in range(0, 360, 60):
        leaf = pin(top_left, outlined_petal(size, line_size, color))
        result = compose(result, rotate(angle, leaf))
    return result

show_graphic(ring(50, 5, hsl_color(HUE, SATURATION, 0.5)))
```

```python
#teacher-only
def flower(size: float) -> Graphic:
    line_size = size * 0.03
    result = empty_graphic()
    for percent in range(10, 100, 10):
        ri = ring(size * percent / 100, line_size, hsl_color(HUE, SATURATION, 0.75 - percent / 200))
        result = compose(result, ri)
    return result

show_graphic(flower(100))

from pytamaro import save_graphic
save_graphic("nested-flower.png", flower(300))
```
