```python
#teacher-only
from pytamaro import Graphic, rgb_color, text, rectangle, empty_graphic, overlay, rotate, beside


TEXT_COLOR = rgb_color(226, 191, 152)
TEXT_BACKGROUND = rgb_color(80, 58, 47)


def letter_row(square_size: float, top: bool) -> Graphic:
    result = empty_graphic()
    for r in ' abcdefgh ':
        label = text(r, "Arial", square_size / 2, TEXT_COLOR)
        cell = overlay(
            rotate(180, label) if top else label,
            rectangle(square_size, square_size, TEXT_BACKGROUND),
        )
        result = beside(result, cell)
    return result
```

```python
#teacher-only
from pytamaro import Graphic, rgb_color, rectangle, empty_graphic, beside, above, show_graphic


WHITE = rgb_color(226, 191, 152)
BLACK = rgb_color(157, 94, 58)


def chessboard_body(square_size: float) -> Graphic:
    result = empty_graphic()
    for r in range(8):
        row = empty_graphic()
        for c in range(8):
            color = WHITE if (r + c) % 2 == 0 else BLACK
            cell = rectangle(square_size, square_size, color)
            row = beside(row, cell)
        result = above(result, row)
    return result

show_graphic(chessboard_body(80))
```

```python
#teacher-only
from pytamaro import Graphic, above, beside, show_graphic


def chessboard(square_size: float) -> Graphic:
    return above(
        letter_row(square_size, True),
        above(
            beside(
                number_col(square_size, True),
                beside(
                    chessboard_body(square_size),
                    number_col(square_size, False),
                ),
            ),
            letter_row(square_size, False),
        ),
    )


show_graphic(chessboard(80))
```
