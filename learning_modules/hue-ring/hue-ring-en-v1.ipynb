```python
#teacher-only
from pytamaro import (
    Graphic, 
    ellipse, empty_graphic, white,
    rotate, overlay, pin, compose, bottom_left, 
    show_graphic
)


def hue_ring(outer_radius: float, inner_radius: float) -> Graphic:
    result = empty_graphic()
    for i in range(360):
        result = compose(result, rotate(i, pin(bottom_left, hue_to_sector(outer_radius, i))))
    return overlay(ellipse(inner_radius * 2, inner_radius * 2, white), result)


show_graphic(hue_ring(150, 120))
```
