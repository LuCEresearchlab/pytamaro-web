### Compose a Ring of Seeds

Create a function `ring_of_seeds` that produces a ring of `count` seeds, evenly spread across the 360 degrees of a circle.
The inside diameter of the ring should correspond to `diameter`
(the seeds should form a ring just outside the diameter).
Repetitively call `pitaya_seed` to produce the ring of seeds.

Hint: Place a rectangle beside a seed and pin the result on the edge of the rectangle,
then rotate that.

```python
from pytamaro import empty_graphic, rotate, pin, beside, rectangle, black, center_left, compose

def ring(count: int, ringDiameter: float, seed: Graphic) -> Graphic:
    result = empty_graphic()
    for i in range(count):
        angle = 360.0 / count * i
        s = rotate(
            angle,
            pin(
                center_left,
                beside(
                    rectangle(ringDiameter / 2, 0, black),
                    seed
                )
            )
        )
        result = compose(result, s)
    return result

show_graphic(ring(8, 100, pitaya_seed(20)))
```
