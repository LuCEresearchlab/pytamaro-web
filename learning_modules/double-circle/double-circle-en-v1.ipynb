```python
#teacher-only

from pytamaro import (
    Graphic,
    Color,
    rgb_color,
    ellipse,
    rectangle,
    beside,
    above,
    overlay,
    show_graphic,
    save_graphic
)


LEFT_COLOR = rgb_color(128, 35, 88)
#RIGHT_COLOR = rgb_color(244, 101, 0)
RIGHT_COLOR = rgb_color(203, 40, 15)


# TODO


def double_circle(diameter: float) -> Graphic:
    return overlay(
        beside(
            ellipse(diameter, diameter, LEFT_COLOR),
            ellipse(diameter, diameter, RIGHT_COLOR)
        ),
        above(
            rectangle(diameter, diameter / 2, LEFT_COLOR),
            rectangle(diameter, diameter / 2, RIGHT_COLOR)
        )
    )

show_graphic(double_circle(100))

save_graphic("double-circle.svg", double_circle(100))
```
