## Konstruiere ein Margritli (Gänseblümchen)

In dieser Übung wirst Du mit Hilfe der [PyTamaro](https://pytamaro.readthedocs.io) Bibliothek ein Gänseblümchen zeichnen.

Ein Gänseblümchen besteht aus weissen Blättern und einem gelben Blütenstand.

Die Blätter sind in einer Rosette (alle ums Zentrum herum) arrangiert.

### Konstruiere ein Blatt

```python
from pytamaro.de import Grafik, weiss, ellipse, fixiere, drehe

def blatt(laenge: float) -> Grafik:
    return ellipse(laenge, laenge / 2, weiss)
```

Zeige ein Blatt an:

```python
from pytamaro.de import zeige_grafik

zeige_grafik(blatt(200))
```

### Blütenstand

Entwickle eine Funktion, die einen Blütenstand mit gegebenem Durchmesser erzeugt:

```python
from pytamaro.de import Grafik, rgb_farbe, ellipse

def bluetenstand(durchmesser: float) -> Grafik:
    GELB = rgb_farbe(255, 206, 46)
    return ellipse(durchmesser, durchmesser, GELB)
```

Erzeuge einen Blütenstand mit Radius 100 und zeige ihn an:

```python
zeige_grafik(bluetenstand(100))
```

### Blattrosette

Schreibe eine Funktion `blattrosette` damit diese eine Rosette aus `anzahl` Blättern erzeugt.

Verwende `fixiere`, `drehe`, `kombiniere` und eine Schleife, um die Blätter zu erzeugen, zu fixieren, und dann mit drehen und kombinieren in eine Rosette zu arrangieren.

```python
from pytamaro.de import leere_grafik, kombiniere

def blattrosette(durchmesser: float, anzahl: int) -> Grafik:
    resultat = leere_grafik()
    for i in range(anzahl):
        winkel = 360 * i / anzahl
        resultat = kombiniere(resultat, drehe(winkel, fixiere("links", "mitte", blatt(durchmesser / 2))))
    return resultat
```

Erstelle eine Blattrosette mit Durchmesser 200 und 11 Blättern:

```python
zeige_grafik(blattrosette(200, 11))
```

### Gaenseblümchen

Implementiere die Funktion `gaensebluemchen`, welche ein Gänseblümchen konstruiert.

Dazu kombiniere den Blütenstand mit dem Blätterkranz.

```python
def gaensebluemchen(durchmesser: float, anzahl_blaetter: int) -> Grafik:
    return kombiniere(bluetenstand(durchmesser / 3), blattrosette(durchmesser, anzahl_blaetter))

zeige_grafik(gaensebluemchen(200, 11))
```
