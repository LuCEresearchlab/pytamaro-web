```python
from pytamaro import (
    Color, 
    Graphic, 
    above, 
    triangle, 
    rectangle, 
    red, 
    white,
    show_graphic)


def house(width: float, roof_color: Color, wall_color: Color) -> Graphic:
    assert width >= 0
    return above(triangle(width, roof_color), rectangle(width, width, wall_color))

def standard_house(width: float) -> Graphic:
    assert width >= 0
    return house(width, red, white)


# code that shows result
show_graphic(standard_house(100))
```
